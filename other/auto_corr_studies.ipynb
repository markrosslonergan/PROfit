```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import acf
from matplotlib.backends.backend_pdf import PdfPages

filename = '/home/mark/work/PROfit_Dev/working_dir/MCMC/testMCMC'  # Replace with your file path
# its a 
df = pd.read_csv(filename, delim_whitespace=True, header=0)
N = df.shape[1]

with PdfPages('mcmc_diagnostics.pdf') as pdf:

    # Hairy Catapillar plots
    fig, axes = plt.subplots(N, 1, figsize=(8, 2*N), sharex=True)
    for i in range(N):
        axes[i].plot(df.iloc[:, i])
        axes[i].set_title(df.columns[i])
    axes[-1].set_xlabel('Sample')
    fig.suptitle('Trace plots (hairy caterpillars)', y=1.02)
    plt.tight_layout()
    pdf.savefig(fig)
    plt.close(fig)

    # Individual autocorrelation plots
    max_lag = 50
    fig, axes = plt.subplots(N, 1, figsize=(8, 2*N))
    for i in range(N):
        acorr = acf(df.iloc[:, i], nlags=max_lag, fft=True)
        axes[i].stem(range(len(acorr)), acorr)
        axes[i].set_ylabel('ACF')
        axes[i].set_title(f'Autocorrelation: {df.columns[i]}')
    axes[-1].set_xlabel('Lag')
    plt.tight_layout()
    pdf.savefig(fig)
    plt.close(fig)

    #  Combined autocorrelation line plot
    fig, ax = plt.subplots(figsize=(10, 6))
    for i in range(N):
        acorr = acf(df.iloc[:, i], nlags=max_lag, fft=True)
        ax.plot(range(len(acorr)), acorr, label=df.columns[i])
    ax.set_xlabel('Lag')
    ax.set_ylabel('Autocorrelation')
    ax.set_title('Autocorrelation Comparison Across Parameters')
    ax.legend(loc='upper right', fontsize='small', ncol=2)
    ax.grid(True)
    plt.tight_layout()
    pdf.savefig(fig)
    plt.close(fig)

print("Saved plots to mcmc_diagnostics.pdf")
```
